```python
#IMPORTING PANDAS AND NUMPY LIBRARY
import pandas as pd
import numpy as np
```

```python
#FETCHING THE CSV FILE
df=pd.read_csv('customer_shopping_behavior.csv')
df.head()
```

```python
#FETCHING INFORMATION ABOUT COLUMN NAME
df.info()
```

```python
#FETCHING HOW MANY NULL VALUES ARE THERE
df.isnull().sum()
```

```python
#FILLING NA VALUES WITH MEDIAN VALUES
df['Review Rating']=df.groupby('Category')['Review Rating'].transform (lambda a:a.fillna(a.median()))
df.isnull().sum()
```

```python
#REPLACING SPACES WITH _
df.columns=df.columns.str.replace(' ','_')
df.columns
```

```python
#CREATING A COLUMN age_group
labels=['Young Adult','Adult','Middle-aged','Senior']
df['Age_group']=pd.qcut(df['Age'],q=4,labels=labels)
```

```python
df[['Age','Age_group']].head()
```

```python

#creating a column frequency of purchases
df['Frequency_of_Purchases'].unique()
```

```python
f_m ={'Fortnightly': 14, 'Weekly': 7, 'Annually': 365, 'Quarterly': 90, 'Bi -weekly': 14}
df['Frequency_of_Purchases_days'] = df['Frequency_of_Purchases'].map(f_m)
df
```

```python
(df['Promo_Code_Used']==df['Discount_Applied']).all()
```

```python
df.info()
```

```python
#DRPPING promo_code_used COLUMN
df = df.drop('Promo_Code_Used', axis=1)
df.columns
```

```python
df.rename(columns={'Purchase_Amount_(USD)':'Purchase_Amount'},inplace=True)
```
